# Câu 37:
Viết thuật toán cho phương pháp lặp Gauss-Seidel giải hệ phương trình $AX = B$ với sai số $\varepsilon$ cho trước.

## Phương pháp lặp Gauss-Seidel

### Bài toán

Cho hệ phương trình tuyến tính:

$$Ax = b$$

với $A \in \text{mat}(n, \mathbb{R})$ khả nghịch, $b \in \mathbb{R}^n$, sai số cho trước $\varepsilon > 0$.

**Yêu cầu:** Tìm nghiệm xấp xỉ $x^* \in \mathbb{R}^n$ bằng cách tách $A = D - L - U$, trong đó:

- $D$ — ma trận đường chéo của $A$
- $L$ — phần tam giác dưới nghiêm ngặt (đổi dấu)
- $U$ — phần tam giác trên nghiêm ngặt (đổi dấu)

Công thức lặp Gauss-Seidel dùng ngay các thành phần **vừa cập nhật** trong cùng một vòng:

$$x_i^{(k+1)} = \frac{1}{a_{ii}} \left( b_i - \sum_{j < i} a_{ij} x_j^{(k+1)} - \sum_{j > i} a_{ij} x_j^{(k)} \right), \quad i = 1, \ldots, n$$

Ở dạng ma trận: $\;x^{(k+1)} = (D - L)^{-1} \big( U x^{(k)} + b \big)$, ma trận lặp $T = (D-L)^{-1} U$.

---

### Điều kiện hội tụ

**Điều kiện đủ dễ kiểm tra:** nếu tồn tại **một** chuẩn ma trận sao cho $\|T\| < 1$ thì phương pháp hội tụ với mọi $x^{(0)}$. Trên thực tế thường kiểm qua tính chất của $A$:

$$A \text{ chéo trội nghiêm ngặt: } \quad |a_{ii}| > \sum_{j \neq i} |a_{ij}|, \quad \forall i$$

hoặc $A$ đối xứng xác định dương. Khi đó Gauss-Seidel **luôn hội tụ**, thường **nhanh hơn Jacobi**.

---

### Thuật toán

**Đầu vào:** Ma trận $A$, vector $b$, sai số $\varepsilon > 0$, số vòng lặp tối đa $N$

**Đầu ra:** Nghiệm xấp xỉ $x^*$ hoặc thông báo không hội tụ

---

**Bước 1 — Kiểm tra tiền đề**

&emsp;Nếu có $a_{ii} = 0$: hoán vị hàng để đường chéo khác 0.

**Bước 2 — Khởi tạo**

&emsp;Chọn $x^{(0)} \in \mathbb{R}^n$ (thường $x^{(0)} = \mathbf{0}$). Đặt $k \leftarrow 0$.

**Bước 3 — Vòng lặp:** lặp khi $k < N$

&emsp;**3.1.** Với mỗi $i = 1, \ldots, n$ (theo thứ tự tăng), cập nhật **tại chỗ**:

$$x_i \leftarrow \frac{1}{a_{ii}} \Big( b_i - \sum_{j < i} a_{ij} x_j - \sum_{j > i} a_{ij} x_j \Big)$$

&emsp;&emsp;*(Tổng $\sum_{j<i}$ đã dùng giá trị mới cập nhật trong chính vòng này.)*

&emsp;**3.2.** Sai số: $\delta = \| x^{(k+1)} - x^{(k)} \|_\infty$

&emsp;**3.3.** Nếu $\delta < \varepsilon$: **dừng**, trả về $x^{(k+1)}$

&emsp;**3.4.** $k \leftarrow k + 1$

**Bước 4 — Kết thúc**

&emsp;Nếu $k = N$ mà chưa hội tụ: thông báo *"Vượt quá số vòng lặp tối đa"*.

---

### Lưu ý

- **Gauss-Seidel vs Jacobi:** Jacobi tính toàn bộ $x^{(k+1)}$ từ $x^{(k)}$ cũ (dễ song song). Gauss-Seidel dùng ngay giá trị vừa cập nhật ⟹ **hội tụ nhanh hơn** (thường gấp đôi tốc độ) nhưng **tuần tự**, không song song hoá được theo thành phần.

- **Cập nhật tại chỗ:** Vì dùng giá trị mới ngay, có thể ghi đè trực tiếp lên vector $x$ — không cần lưu riêng $x^{(k)}$ (chỉ cần bản sao để đo sai số).

- **Đánh giá sai số:** $\delta = \|x^{(k+1)} - x^{(k)}\|_\infty$ là **sai số tuyệt đối** giữa hai bước. Muốn **sai số tương đối** so với nghiệm: $\|x - x^*\| / \|x^*\|$ (cần biết $x^*$, hoặc ước lượng qua residual $\|Ax - b\|$).

- **Thứ tự phương trình ảnh hưởng hội tụ:** Khác Jacobi, kết quả Gauss-Seidel phụ thuộc thứ tự duyệt $i$. Sắp hàng sao cho chéo trội mạnh giúp hội tụ nhanh hơn.

# Câu 38:
Dùng phương pháp lặp Gauss-Seidel tìm nghiệm gần đúng của phương trình $(A + aI)x = b$ với sai số tuyệt đối không vượt quá $10^{-5}$, biết $a$ là số thứ tự theo danh sách thi của bạn; $I$ là ma trận đơn vị cùng cấp với $A$; ma trận $A$, vector $b$ cho trong file **GK20222** với pass mở file là **`gts2023gts`**.

Thực hiện thuật toán theo phương pháp đã chọn để tìm nghiệm theo yêu cầu. Đánh giá sai số tương đối cho nghiệm tìm được.

> *Ghi chú:* Ở đây lấy $a = 20$ (số thứ tự danh sách thi); ma trận $A$ và vector $b$ được **tự cho** (chéo trội, cấp 6) để minh hoạ thuật toán.

```python
import numpy as np


# ── Định dạng số khoa học kiểu 1.23×10⁻⁵ ─────────────────────────
def _sci(x, sig=4):
    if abs(x) < 1e-14:
        return "0"
    s = f"{x:.{sig}e}"
    m, e = s.split("e")
    exp = int(e)
    sup = str(abs(exp)).translate(str.maketrans("0123456789", "⁰¹²³⁴⁵⁶⁷⁸⁹"))
    sign = "⁻" if exp < 0 else ""
    return f"{m}×10{sign}{sup}"


def gauss_seidel(A, b, eps=1e-5, max_iter=1000, log=6):
    """
    Phương pháp lặp Gauss-Seidel: Ax = b

    Tham số:
        A, b    : ma trận hệ số và vế phải
        eps     : sai số tuyệt đối cho phép (‖x^{k+1}-x^k‖∞)
        max_iter: số vòng lặp tối đa
        log     : in log vài vòng đầu (0 = tắt)

    Trả về:
        x, số vòng lặp, thông báo
    """
    A, b = np.array(A, dtype=float), np.array(b, dtype=float)
    n = len(b)
    diag = np.diag(A)
    if np.any(diag == 0):
        return None, 0, "Phần tử đường chéo = 0, cần hoán vị hàng"

    # Điều kiện đủ: chéo trội nghiêm ngặt?
    doi = np.sum(np.abs(A - np.diag(diag)), axis=1)
    cheo_troi = np.all(np.abs(diag) > doi)

    x = np.zeros(n)
    if log:
        print(f"A chéo trội nghiêm ngặt: {cheo_troi}  →  hội tụ đảm bảo\n")
        print(f"{'k':>3}  {'x':>54}  {'sai số':>12}")
        print(f"{'0':>3}  {np.array2string(x, precision=6):>54}")

    for k in range(1, max_iter + 1):
        x_old = x.copy()
        for i in range(n):
            s1 = A[i, :i] @ x[:i]              # dùng x MỚI (đã cập nhật)
            s2 = A[i, i+1:] @ x_old[i+1:]      # dùng x CŨ
            x[i] = (b[i] - s1 - s2) / A[i, i]
        delta = np.max(np.abs(x - x_old))     # sai số tuyệt đối
        if log and k <= log:
            print(f"{k:>3}  {np.array2string(x, precision=6):>54}  {_sci(delta):>12}")
        if delta < eps:
            return x, k, "ok"
    return x, max_iter, "Vượt quá số vòng lặp tối đa"


# ====== Áp dụng: (A + aI)x = b với a = 20 ======
if __name__ == "__main__":
    # A và b tự cho (chéo trội, cấp 6)
    A0 = np.array([
        [ 8., -1.,  2.,  0.,  1., -1.],
        [ 1.,  9., -1.,  2.,  0.,  1.],
        [ 2.,  1., 10., -1.,  1.,  0.],
        [ 0.,  2., -1., 12.,  1.,  1.],
        [ 1.,  0.,  1.,  2.,  9., -1.],
        [-1.,  1.,  0.,  1., -1., 11.],
    ])
    b = np.array([7., 12., 15., 20., 9., 10.])
    a = 20                                    # số thứ tự theo danh sách thi
    A = A0 + a * np.eye(len(A0))              # ma trận A + aI

    x, k, info = gauss_seidel(A, b, eps=1e-5)
    print(f"\nTrạng thái : {info}")
    print(f"Số vòng lặp: {k}")
    print("Nghiệm x   : [" + "  ".join(f"{v:.8f}" for v in x) + "]")

    # ── Đánh giá sai số ──
    x_true  = np.linalg.solve(A, b)                       # nghiệm đúng để đối chiếu
    abs_err = np.max(np.abs(x - x_true))                  # sai số tuyệt đối
    rel_err = np.linalg.norm(x - x_true) / np.linalg.norm(x_true)  # sai số tương đối
    res     = np.linalg.norm(A @ x - b)                   # residual

    print("\n=== Đánh giá sai số ===")
    print(f"  Sai số tuyệt đối  ‖x - x*‖∞      = {_sci(abs_err)}   (≤ 10⁻⁵ ✓)")
    print(f"  Sai số tương đối  ‖x - x*‖/‖x*‖  = {_sci(rel_err)}")
    print(f"  Residual          ‖A x - b‖      = {_sci(res)}")
```

# Câu 39:
Viết thuật toán cho **phương pháp lặp Seidel** giải hệ ở dạng $x = Bx + d$.

## Phương pháp lặp Seidel (dạng $x = Bx + d$)

### Bài toán

Cho hệ đã đưa về **dạng lặp**:

$$x = Bx + d$$

với $B \in \text{mat}(n, \mathbb{R})$, $d \in \mathbb{R}^n$, sai số cho trước $\varepsilon > 0$.

**Ý tưởng.** Tách ma trận lặp $B$ thành hai phần:

$$B = B_1 + B_2$$

- $B_1$ — phần **tam giác dưới nghiêm ngặt** của $B$ (các phần tử $b_{ij}$ với $j < i$),
- $B_2$ — phần **tam giác trên** của $B$, **gồm cả đường chéo** (các phần tử $b_{ij}$ với $j \ge i$).

Khác với Jacobi (dùng toàn bộ $x^{(k)}$ cũ), Seidel **dùng ngay** các thành phần vừa được cập nhật trong cùng một vòng. Công thức lặp:

$$x^{(k+1)} = B_1\, x^{(k+1)} + B_2\, x^{(k)} + d, \qquad k \ge 0$$

Vì $B_1$ là tam giác dưới nghiêm ngặt nên $x_i^{(k+1)}$ chỉ phụ thuộc các thành phần $x_j^{(k+1)}$ với $j < i$ (đã tính xong trước đó) — hệ **giải được tuần tự** không cần nghịch đảo:

$$x_i^{(k+1)} = \sum_{j < i} b_{ij}\, x_j^{(k+1)} \;+\; \sum_{j \ge i} b_{ij}\, x_j^{(k)} \;+\; d_i, \qquad i = 1, \ldots, n$$

Ở dạng ma trận, chuyển $B_1 x^{(k+1)}$ sang vế trái:

$$(I - B_1)\, x^{(k+1)} = B_2\, x^{(k)} + d
\;\;\Longrightarrow\;\;
x^{(k+1)} = (I - B_1)^{-1}\big(B_2\, x^{(k)} + d\big)$$

Ma trận lặp tương đương $T = (I - B_1)^{-1} B_2$.

---

### Điều kiện hội tụ

**Điều kiện đủ dễ kiểm tra:** nếu $\|B\|_\infty < 1$ (hoặc $\|B\|_1 < 1$) thì Seidel hội tụ với mọi $x^{(0)}$, và thường **nhanh hơn Jacobi**. Tổng quát hơn, chỉ cần tồn tại **một** chuẩn ma trận sao cho $\|T\| < 1$ với $T = (I - B_1)^{-1} B_2$ là đủ để kết luận hội tụ.

---

### Thuật toán

**Đầu vào:** Ma trận $B$, vector $d$, sai số $\varepsilon > 0$, số vòng lặp tối đa $N$

**Đầu ra:** Nghiệm xấp xỉ $x^*$ hoặc thông báo không hội tụ

---

**Bước 1 — Khởi tạo**

&emsp;Chọn $x^{(0)} \in \mathbb{R}^n$ (thường $x^{(0)} = \mathbf{0}$). Đặt $k \leftarrow 0$.

**Bước 2 — Vòng lặp:** lặp khi $k < N$

&emsp;**2.1.** Với mỗi $i = 1, \ldots, n$ (theo thứ tự tăng), cập nhật **tại chỗ**:

$$x_i \leftarrow \sum_{j < i} b_{ij}\, x_j \;+\; \sum_{j \ge i} b_{ij}\, x_j^{\text{cũ}} \;+\; d_i$$

&emsp;&emsp;*(Tổng $\sum_{j<i}$ dùng ngay giá trị $x_j$ vừa cập nhật trong chính vòng này — đó là $B_1$; tổng $\sum_{j \ge i}$ dùng giá trị cũ — đó là $B_2$.)*

&emsp;**2.2.** Sai số: $\delta = \| x^{(k+1)} - x^{(k)} \|_\infty$

&emsp;**2.3.** Nếu $\delta < \varepsilon$: **dừng**, trả về $x^{(k+1)}$

&emsp;**2.4.** $k \leftarrow k + 1$

**Bước 3 — Kết thúc**

&emsp;Nếu $k = N$ mà chưa hội tụ: thông báo *"Vượt quá số vòng lặp tối đa"*.

---

### Lưu ý

- **Quan hệ với Gauss-Seidel:** Nếu hệ gốc là $Ax = b$ với $A = D - L - U$, đặt $B = D^{-1}(L + U)$, $d = D^{-1}b$ thì phương pháp Seidel trên dạng $x = Bx + d$ **trùng** với Gauss-Seidel ở Câu 37. Ở đây $B_1$ ứng với $D^{-1}L$, còn $B_2$ ứng với $D^{-1}U$.

- **Không cần nghịch đảo:** Do $B_1$ tam giác dưới nghiêm ngặt, việc "giải" $(I - B_1)x^{(k+1)} = \dots$ chỉ là thay thế tiến (forward substitution), thực hiện gọn bằng cập nhật tại chỗ theo thứ tự $i$ tăng dần.

- **Cập nhật tại chỗ:** Vì dùng giá trị mới ngay, có thể ghi đè trực tiếp lên vector $x$ — chỉ cần một bản sao $x^{\text{cũ}}$ để đo sai số $\delta$.

```python
import numpy as np


# ── Định dạng số khoa học kiểu 1.23×10⁻⁵ ─────────────────────────
def _sci(x, sig=4):
    if abs(x) < 1e-14:
        return "0"
    s = f"{x:.{sig}e}"
    m, e = s.split("e")
    exp = int(e)
    sup = str(abs(exp)).translate(str.maketrans("0123456789", "⁰¹²³⁴⁵⁶⁷⁸⁹"))
    sign = "⁻" if exp < 0 else ""
    return f"{m}×10{sign}{sup}"


def seidel(B, d, eps=1e-5, max_iter=1000, log=6):
    """
    Phương pháp lặp Seidel cho hệ dạng  x = B x + d.

    Tách B = B1 + B2:
        B1 = tam giác dưới nghiêm ngặt của B   (j < i)
        B2 = tam giác trên của B, gồm đường chéo (j >= i)
    Công thức:  x^{k+1} = B1 x^{k+1} + B2 x^k + d
    (B1 dùng x MỚI, B2 dùng x CŨ → cập nhật tại chỗ, không cần nghịch đảo).

    Tham số:
        B, d    : ma trận lặp và vế phải
        eps     : sai số tuyệt đối cho phép (‖x^{k+1}-x^k‖∞)
        max_iter: số vòng lặp tối đa
        log     : in log vài vòng đầu (0 = tắt)

    Trả về:
        x, số vòng lặp, thông báo
    """
    B, d = np.array(B, dtype=float), np.array(d, dtype=float)
    n = len(d)

    # Điều kiện đủ hội tụ: ‖B‖∞ < 1 ?
    normB = np.max(np.sum(np.abs(B), axis=1))
    hoi_tu = normB < 1

    x = np.zeros(n)
    if log:
        print(f"‖B‖∞ = {normB:.4f}  <  1 : {hoi_tu}  →  hội tụ đảm bảo\n")
        print(f"{'k':>3}  {'x':>54}  {'sai số':>12}")
        print(f"{'0':>3}  {np.array2string(x, precision=6):>54}")

    for k in range(1, max_iter + 1):
        x_old = x.copy()
        for i in range(n):
            s1 = B[i, :i] @ x[:i]             # B1: dùng x MỚI (đã cập nhật)
            s2 = B[i, i:] @ x_old[i:]         # B2: dùng x CŨ (gồm đường chéo)
            x[i] = s1 + s2 + d[i]
        delta = np.max(np.abs(x - x_old))     # sai số tuyệt đối
        if log and k <= log:
            print(f"{k:>3}  {np.array2string(x, precision=6):>54}  {_sci(delta):>12}")
        if delta < eps:
            return x, k, "ok"
    return x, max_iter, "Vượt quá số vòng lặp tối đa"


# ====== Áp dụng: hệ x = B x + d (B tự cho, ‖B‖∞ < 1) ======
if __name__ == "__main__":
    B = np.array([
        [ 0.10, -0.20,  0.05,  0.00],
        [ 0.15,  0.10, -0.10,  0.05],
        [-0.05,  0.20,  0.10, -0.15],
        [ 0.10, -0.05,  0.15,  0.10],
    ])
    d = np.array([1.0, 2.0, -1.0, 0.5])

    x, k, info = seidel(B, d, eps=1e-5)
    print(f"\nTrạng thái : {info}")
    print(f"Số vòng lặp: {k}")
    print("Nghiệm x   : [" + "  ".join(f"{v:.8f}" for v in x) + "]")

    # ── Đánh giá sai số ──
    n = len(d)
    x_true  = np.linalg.solve(np.eye(n) - B, d)           # nghiệm đúng của (I-B)x = d
    abs_err = np.max(np.abs(x - x_true))                  # sai số tuyệt đối
    rel_err = np.linalg.norm(x - x_true) / np.linalg.norm(x_true)  # sai số tương đối
    res     = np.linalg.norm((np.eye(n) - B) @ x - d)     # residual ‖(I-B)x - d‖

    print("\n=== Đánh giá sai số ===")
    print(f"  Sai số tuyệt đối  ‖x - x*‖∞      = {_sci(abs_err)}   (≤ 10⁻⁵ ✓)")
    print(f"  Sai số tương đối  ‖x - x*‖/‖x*‖  = {_sci(rel_err)}")
    print(f"  Residual          ‖(I-B)x - d‖   = {_sci(res)}")
```
