# Câu 43:
Viết thuật toán tìm đa thức đặc trưng của ma trận $A$ bằng phương pháp Danilevski. Áp dụng với ma trận cấp 5

## Phương pháp Danilevski

### Bài toán

Cho ma trận $A \in \text{mat}(n, \mathbb{R})$. Tìm đa thức đặc trưng:

$$P(\lambda) = \det(A - \lambda I) = (-1)^n \left( \lambda^n - p_1 \lambda^{n-1} - p_2 \lambda^{n-2} - \cdots - p_n \right)$$

**Ý tưởng:** Đưa $A$ về **dạng chuẩn Frobenius** $P$ (đồng dạng với $A$) bằng một dãy phép biến đổi đồng dạng. Ma trận Frobenius có dạng:

$$
P = \begin{bmatrix}
p_1 & p_2 & p_3 & \cdots & p_{n-1} & p_n \\
1 & 0 & 0 & \cdots & 0 & 0 \\
0 & 1 & 0 & \cdots & 0 & 0 \\
\vdots & & \ddots & & & \vdots \\
0 & 0 & 0 & \cdots & 1 & 0
\end{bmatrix}
$$

Vì $P$ đồng dạng với $A$ nên chúng có **cùng đa thức đặc trưng**, và hàng đầu của $P$ chính là các hệ số:

$$P(\lambda) = (-1)^n \left( \lambda^n - p_1 \lambda^{n-1} - p_2 \lambda^{n-2} - \cdots - p_n \right)$$

---

### Cơ sở của phép biến đổi

Ta khử dần từng hàng, đi từ hàng $n$ lên hàng $2$. Ở bước thứ $k$ (đưa hàng $k$ về dạng $[0, \ldots, 0, 1, 0, \ldots, 0]$ — số $1$ ở cột $k-1$), ta dùng phần tử **trụ** $a_{k,k-1}$ và ma trận biến đổi $M_{k-1}$ có dạng: ma trận đơn vị nhưng **hàng $k-1$** được thay bằng:

$$M_{k-1}[k-1, j] =
\begin{cases}
\dfrac{1}{a_{k,k-1}} & j = k-1 \\[2mm]
-\dfrac{a_{k,j}}{a_{k,k-1}} & j \neq k-1
\end{cases}$$

Ma trận nghịch đảo $M_{k-1}^{-1}$ cũng là ma trận đơn vị, nhưng hàng $k-1$ thay bằng chính hàng $k$ của $A$ hiện tại: $M_{k-1}^{-1}[k-1, j] = a_{k,j}$.

Phép biến đổi đồng dạng ở mỗi bước:

$$A^{(k-1)} = M_{k-1}^{-1} \, A^{(k)} \, M_{k-1}$$

giữ nguyên đa thức đặc trưng và đưa thêm một hàng về dạng chuẩn.

---

### Thuật toán

**Đầu vào:** Ma trận $A$ vuông cấp $n$

**Đầu ra:** Các hệ số $p_1, p_2, \ldots, p_n$ của đa thức đặc trưng

---

**Bước 1 — Khởi tạo**

&emsp;Đặt $A^{(n)} \leftarrow A$, và $k \leftarrow n$ (hàng đang xử lý)

**Bước 2 — Vòng lặp khử:** với $k = n, n-1, \ldots, 2$

&emsp;**2.1.** Xét phần tử trụ $a_{k,k-1}$ của ma trận hiện tại.

&emsp;**2.2.** *(Xử lý trụ suy biến)* Nếu $a_{k,k-1} \approx 0$:

&emsp;&emsp;• Tìm cột $j < k-1$ sao cho $a_{k,j} \neq 0$ rồi **hoán vị** đồng thời cột $j \leftrightarrow k-1$ và hàng $j \leftrightarrow k-1$ (giữ tính đồng dạng), sau đó tiếp tục.

&emsp;&emsp;• Nếu $a_{k,j} = 0$ với mọi $j < k-1$: ma trận **tách khối** — tách bài toán thành hai đa thức đặc trưng con nhỏ hơn rồi nhân lại.

&emsp;**2.3.** Lập ma trận biến đổi $M_{k-1}$ và nghịch đảo $M_{k-1}^{-1}$ theo công thức ở trên.

&emsp;**2.4.** Cập nhật: $\;A \leftarrow M_{k-1}^{-1} \, A \, M_{k-1}$

&emsp;&emsp;*(Sau bước này hàng $k$ có dạng $[0,\ldots,0,\,1,\,0,\ldots,0]$)*

**Bước 3 — Trích hệ số**

&emsp;Sau khi $k = 2$ hoàn tất, ma trận đã ở dạng Frobenius. **Hàng đầu tiên** chính là:

$$[p_1, p_2, \ldots, p_n]$$

**Bước 4 — Kết luận**

&emsp;Đa thức đặc trưng:

$$P(\lambda) = (-1)^n \left( \lambda^n - p_1 \lambda^{n-1} - p_2 \lambda^{n-2} - \cdots - p_n \right)$$

---

### Lưu ý

- **Chi phí tính toán:** Danilevski chỉ tốn $O(n^3)$ phép tính — nhanh hơn nhiều so với khai triển định thức $\det(A - \lambda I)$ trực tiếp ($O(n!)$), và nhanh hơn phương pháp Leverrier–Faddeev ($O(n^4)$).

- **Trụ suy biến $a_{k,k-1} = 0$:** Đây là điểm khó nhất của phương pháp. Cần xử lý theo hai trường hợp ở Bước 2.2 (hoán vị hoặc tách khối), nếu không thuật toán sẽ chia cho $0$.

- **Tính đồng dạng $A \sim P$:** Vì $P = S^{-1} A S$ với $S = M_{n-1} M_{n-2} \cdots M_1$, ma trận $S$ còn cho phép **khôi phục vector riêng** của $A$ từ vector riêng của $P$ (vốn có dạng Vandermonde đơn giản theo $\lambda$).

- **Sai số làm tròn:** Khi $|a_{k,k-1}|$ nhỏ (dù khác $0$), phép chia khuếch đại sai số. Nên chọn trụ có trị tuyệt đối lớn nhất trong hàng để hoán vị (**pivot từng phần**) nhằm ổn định số học.

- **Kiểm tra kết quả:** Có thể đối chiếu $p_1 = \text{tr}(A)$ (vết ma trận) và $(-1)^n p_n = \det(A)$ để xác nhận hệ số tính đúng.

```python
import numpy as np


# ── Định dạng số khoa học kiểu 1.23×10⁻⁵ ─────────────────────────
def _sci(x, sig=4):
    if abs(x) < 1e-14:
        return "0"
    s = f"{x:.{sig}e}"
    m, e = s.split("e")
    exp = int(e)
    sup = str(abs(exp)).translate(str.maketrans("0123456789", "⁰¹²³⁴⁵⁶⁷⁸⁹"))
    sign = "⁻" if exp < 0 else ""
    return f"{m}×10{sign}{sup}"


def _print_mat(M, label):
    print(f"\n{label}:")
    for row in M:
        print("  " + "  ".join(f"{v:11.5f}" for v in row))


def _poly_str(p, n):
    """In P(λ) = (-1)^n (λ^n - p1 λ^(n-1) - ... - pn)."""
    terms = [f"λ^{n}"]
    for i, pi in enumerate(p):
        power = n - 1 - i
        coef = -pi                       # dấu trước hệ số trong ngoặc
        sign = "-" if coef < 0 else "+"
        mono = "" if power == 0 else (f"λ^{power}" if power > 1 else "λ")
        terms.append(f"{sign} {abs(coef):.4f}{('·' + mono) if mono else ''}")
    inside = " ".join(terms)
    prefix = "" if n % 2 == 0 else "-"    # (-1)^n
    return f"P(λ) = {prefix}( {inside} )"


# ── Thuật toán Danilevski ────────────────────────────────────────
def danilevski(A, log=True):
    """
    Tìm đa thức đặc trưng của A bằng phương pháp Danilevski.

    Trả về:
        p    : mảng hệ số [p1, p2, ..., pn] (hàng đầu ma trận Frobenius)
        F    : ma trận dạng chuẩn Frobenius (đồng dạng với A)
        info : "ok" / thông báo
    """
    A = np.array(A, dtype=float)
    n = len(A)
    A = A.copy()

    # Khử hàng k = n, n-1, ..., 2 → đưa số 1 về cột k-1
    for k in range(n - 1, 0, -1):          # chỉ số 0-based: hàng k, trụ ở cột k-1
        piv = A[k, k - 1]

        # Xử lý trụ suy biến: hoán vị đồng thời hàng/cột j <-> k-1
        if abs(piv) < 1e-12:
            j = next((c for c in range(k - 1) if abs(A[k, c]) > 1e-12), None)
            if j is None:
                return None, A, f"Trụ suy biến tại hàng {k+1} (ma trận tách khối)"
            A[[j, k - 1], :] = A[[k - 1, j], :]
            A[:, [j, k - 1]] = A[:, [k - 1, j]]
            piv = A[k, k - 1]

        # M^{-1} = I nhưng hàng (k-1) là hàng k hiện tại;  M là I nhưng hàng (k-1) đảo ngược
        M = np.eye(n)
        M[k - 1, :] = -A[k, :] / piv
        M[k - 1, k - 1] = 1.0 / piv

        Minv = np.eye(n)
        Minv[k - 1, :] = A[k, :]

        A = Minv @ A @ M

        if log:
            _print_mat(A, f"Sau khi khử hàng {k+1} (trụ = {_sci(piv)})")

    p = A[0].copy()
    return p, A, "ok"


# ====== Áp dụng với ma trận cấp 5 ======
if __name__ == "__main__":
    A = np.array([
        [ 5.0,  2.0,  1.0,  3.0,  1.0],
        [ 1.0,  4.0,  2.0,  1.0,  2.0],
        [ 2.0,  1.0,  6.0,  2.0,  1.0],
        [ 1.0,  3.0,  1.0,  5.0,  2.0],
        [ 2.0,  1.0,  2.0,  1.0,  4.0],
    ])
    n = len(A)

    _print_mat(A, "Ma trận A ban đầu")

    p, F, info = danilevski(A)
    print(f"\nTrạng thái: {info}")

    _print_mat(F, "Dạng chuẩn Frobenius (hàng đầu = hệ số)")

    print("\nHệ số:", "  ".join(f"p{i+1} = {v:.4f}" for i, v in enumerate(p)))
    print("\n" + _poly_str(p, n))

    # ── Kiểm tra kết quả ──
    tr, det = np.trace(A), np.linalg.det(A)
    print("\n=== Kiểm tra ===")
    print(f"  p1        = {p[0]:.6f}   |  tr(A)          = {tr:.6f}   →  {'khớp' if abs(p[0]-tr)<1e-8 else 'lệch'}")
    pn_check = (-1)**n * (-p[-1])        # P(0) = det(A - 0·I) = det(A)
    print(f"  (-1)^n·(-pn) = {pn_check:.6f}   |  det(A)      = {det:.6f}   →  {'khớp' if abs(pn_check-det)<1e-6 else 'lệch'}")

    # Đối chiếu với numpy (hệ số đa thức đặc trưng chuẩn)
    np_coef = np.poly(A)          # [1, -c1, c2, ...] của det(λI - A)
    my_coef = np.concatenate(([1.0], -p))   # λ^n - p1 λ^(n-1) - ...
    err = np.max(np.abs(np_coef - my_coef))
    print(f"  Sai số so với numpy.poly : {_sci(err)}  →  {'ĐÚNG ✓' if err < 1e-6 else 'SAI ✗'}")

    roots = np.roots(my_coef)
    print("\nGiá trị riêng (nghiệm của P(λ)):")
    for r in np.sort_complex(roots):
        print(f"  λ = {r.real:.6f}" + (f" + {r.imag:.6f}i" if abs(r.imag) > 1e-9 else ""))
```

# Câu 44:
Viết thuật toán tìm các giá trị riêng và các véc-tơ riêng tương ứng bằng phương pháp Danielevski. Áp dụng cho một ma trận cấp 6.

## Giá trị riêng & vector riêng bằng Danilevski

### Bài toán

Cho ma trận $A \in \text{mat}(n, \mathbb{R})$. Tìm các **giá trị riêng** $\lambda_i$ và **vector riêng** $x_i$ tương ứng, tức là $A x_i = \lambda_i x_i$.

**Ý tưởng:** Tận dụng kết quả của phương pháp Danilevski (Câu 43): đưa $A$ về dạng chuẩn Frobenius $P$ đồng dạng với $A$, rồi khai thác cấu trúc đơn giản của vector riêng của $P$ để **suy ngược** ra vector riêng của $A$.

---

### Cơ sở lý thuyết

**1) Ma trận biến đổi tích luỹ $S$.** Ở Câu 43, mỗi bước khử hàng $k$ dùng ma trận $M_{k-1}$ sao cho:

$$P = M_1^{-1} \cdots M_{n-2}^{-1} M_{n-1}^{-1} \, A \, M_{n-1} M_{n-2} \cdots M_1 = S^{-1} A S$$

với $S = M_{n-1} M_{n-2} \cdots M_1$ (tích các ma trận biến đổi theo đúng thứ tự khử từ hàng $n$ về hàng $2$).

**2) Giá trị riêng.** Vì $P \sim A$ nên chúng có **cùng đa thức đặc trưng**. Từ hàng đầu $[p_1, \ldots, p_n]$ của $P$:

$$P(\lambda) = (-1)^n \left( \lambda^n - p_1 \lambda^{n-1} - \cdots - p_n \right)$$

Nghiệm của $P(\lambda) = 0$ chính là các giá trị riêng $\lambda_1, \ldots, \lambda_n$.

**3) Vector riêng của $P$ (dạng Frobenius).** Với mỗi giá trị riêng $\lambda$, vector riêng của ma trận Frobenius $P$ có **dạng luỹ thừa** (Vandermonde) rất đơn giản:

$$y = \begin{bmatrix} \lambda^{n-1} & \lambda^{n-2} & \cdots & \lambda & 1 \end{bmatrix}^T$$

*Chứng minh nhanh:* Với $P$ dạng Frobenius, phương trình $P y = \lambda y$ ở các hàng $2, \ldots, n$ cho $y_{i-1} = \lambda\, y_i$, dẫn tới $y_i = \lambda^{n-i}$ (chuẩn hoá $y_n = 1$). Hàng đầu tự động thoả vì $\lambda$ là nghiệm đa thức đặc trưng.

**4) Vector riêng của $A$.** Vì $P = S^{-1} A S$, nếu $P y = \lambda y$ thì:

$$A(S y) = S P y = \lambda (S y) \quad\Longrightarrow\quad \boxed{x = S y}$$

---

### Thuật toán

**Đầu vào:** Ma trận $A$ vuông cấp $n$

**Đầu ra:** Các cặp $(\lambda_i, x_i)$ — giá trị riêng và vector riêng

---

**Bước 1 — Chạy Danilevski, lưu tích biến đổi $S$**

&emsp;Thực hiện khử như Câu 43. Khởi tạo $S \leftarrow I$.

&emsp;Ở mỗi bước khử hàng $k$, sau khi lập $M_{k-1}$: cập nhật $S \leftarrow S \cdot M_{k-1}$.

&emsp;*(Nếu có hoán vị hàng/cột $j \leftrightarrow k-1$ để xử lý trụ suy biến, cần hoán vị cột tương ứng của $S$.)*

**Bước 2 — Giá trị riêng**

&emsp;Lấy hàng đầu $[p_1, \ldots, p_n]$ của $P$, lập đa thức đặc trưng và **giải nghiệm** $\lambda_1, \ldots, \lambda_n$.

**Bước 3 — Vector riêng**

&emsp;Với mỗi $\lambda_i$:

&emsp;&emsp;**3.1.** Lập $y_i = (\lambda_i^{n-1}, \lambda_i^{n-2}, \ldots, \lambda_i, 1)^T$

&emsp;&emsp;**3.2.** Tính $x_i = S\, y_i$

&emsp;&emsp;**3.3.** Chuẩn hoá $x_i \leftarrow x_i / \|x_i\|$

**Bước 4 — Kiểm tra**

&emsp;Xác nhận $\|A x_i - \lambda_i x_i\| \approx 0$ với mỗi $i$.

---

### Lưu ý

- **Chỉ tốn thêm $O(n^2)$ mỗi vector:** Vector riêng của $P$ có sẵn dạng đóng, nên chỉ cần một phép nhân $S y$ — không phải giải lại hệ $(A - \lambda I)x = 0$.

- **Giá trị riêng bội / phức:** Công thức $y = (\lambda^{n-1}, \ldots, 1)^T$ vẫn đúng cho $\lambda$ **phức**. Với **giá trị riêng bội**, ma trận Frobenius chỉ có **một** vector riêng độc lập cho mỗi giá trị riêng (khối Jordan), nên phương pháp không cho đủ vector riêng nếu $A$ không chéo hoá được.

- **Trường hợp tách khối:** Nếu quá trình khử gặp trụ suy biến không hoán vị được (ma trận tách khối), cần xử lý từng khối con rồi ghép — phần cài đặt cơ bản dưới đây giả định không rơi vào trường hợp này.

- **Ổn định số học:** Dạng Vandermonde $(\lambda^{n-1}, \ldots, 1)$ dễ bị **tràn/mất chính xác** khi $|\lambda|$ lớn và $n$ lớn. Có thể chuẩn hoá theo phần tử có trị tuyệt đối lớn nhất trước khi nhân với $S$.

```python
import numpy as np


# ── Định dạng số khoa học kiểu 1.23×10⁻⁵ ─────────────────────────
def _sci(x, sig=4):
    if abs(x) < 1e-14:
        return "0"
    s = f"{x:.{sig}e}"
    m, e = s.split("e")
    exp = int(e)
    sup = str(abs(exp)).translate(str.maketrans("0123456789", "⁰¹²³⁴⁵⁶⁷⁸⁹"))
    sign = "⁻" if exp < 0 else ""
    return f"{m}×10{sign}{sup}"


def _fmt(z):
    """Số thực hoặc phức gọn gàng."""
    if abs(z.imag) < 1e-9:
        return f"{z.real:9.5f}"
    return f"{z.real:9.5f}{'+' if z.imag >= 0 else '-'}{abs(z.imag):.5f}i"


# ── Danilevski: trả về hệ số + ma trận biến đổi tích luỹ S ────────
def danilevski_full(A):
    """
    Đưa A về dạng Frobenius P = S^{-1} A S.

    Trả về:
        p    : hệ số [p1,...,pn] (hàng đầu của P)
        S    : ma trận biến đổi tích luỹ (S = M_{n-1}...M_1)
        info : "ok" / thông báo
    """
    A = np.array(A, dtype=float).copy()
    n = len(A)
    S = np.eye(n)

    for k in range(n - 1, 0, -1):            # khử hàng k, trụ ở cột k-1
        piv = A[k, k - 1]

        if abs(piv) < 1e-12:                 # trụ suy biến → hoán vị
            j = next((c for c in range(k - 1) if abs(A[k, c]) > 1e-12), None)
            if j is None:
                return None, S, f"Trụ suy biến tại hàng {k+1} (ma trận tách khối)"
            A[[j, k - 1], :] = A[[k - 1, j], :]
            A[:, [j, k - 1]] = A[:, [k - 1, j]]
            S[:, [j, k - 1]] = S[:, [k - 1, j]]
            piv = A[k, k - 1]

        M = np.eye(n)
        M[k - 1, :] = -A[k, :] / piv
        M[k - 1, k - 1] = 1.0 / piv

        Minv = np.eye(n)
        Minv[k - 1, :] = A[k, :]

        A = Minv @ A @ M
        S = S @ M                             # tích luỹ biến đổi

    return A[0].copy(), S, "ok"


# ── Giá trị riêng & vector riêng ─────────────────────────────────
def eig_danilevski(A):
    A = np.array(A, dtype=float)
    n = len(A)
    p, S, info = danilevski_full(A)
    if info != "ok":
        return None, None, info

    # Đa thức đặc trưng: λ^n - p1 λ^(n-1) - ... - pn  →  giải nghiệm
    coef = np.concatenate(([1.0], -p))        # [1, -p1, -p2, ..., -pn]
    lambdas = np.roots(coef)

    # Vector riêng: y = (λ^(n-1), ..., λ, 1),  x = S y
    X = np.zeros((n, n), dtype=complex)
    for i, lam in enumerate(lambdas):
        y = np.array([lam ** (n - 1 - j) for j in range(n)], dtype=complex)
        x = S @ y
        X[:, i] = x / np.linalg.norm(x)
    return lambdas, X, "ok"


# ====== Áp dụng với ma trận cấp 6 ======
if __name__ == "__main__":
    A = np.array([
        [ 6.0,  2.0,  1.0,  3.0,  1.0,  2.0],
        [ 1.0,  5.0,  2.0,  1.0,  2.0,  1.0],
        [ 2.0,  1.0,  7.0,  2.0,  1.0,  3.0],
        [ 1.0,  3.0,  1.0,  6.0,  2.0,  1.0],
        [ 2.0,  1.0,  2.0,  1.0,  5.0,  2.0],
        [ 1.0,  2.0,  1.0,  3.0,  1.0,  8.0],
    ])
    n = len(A)

    lambdas, X, info = eig_danilevski(A)
    print(f"Trạng thái: {info}\n")

    print("Giá trị riêng và vector riêng (đã chuẩn hoá):")
    print("─" * 78)
    order = np.argsort(-lambdas.real)
    for idx in order:
        lam = lambdas[idx]
        x = X[:, idx]
        # bỏ phần ảo ~0 khi in vector
        xr = np.array([v.real if abs(v.imag) < 1e-9 else v for v in x], dtype=object)
        res = np.linalg.norm(A @ x - lam * x)
        print(f"λ = {_fmt(lam)}")
        vecstr = "  ".join(f"{v:8.4f}" if isinstance(v, float) else f"{_fmt(v)}" for v in xr)
        print(f"  x = [{vecstr}]")
        print(f"  ||A·x - λ·x|| = {_sci(res)}")
        print("─" * 78)

    # ── Đối chiếu với numpy.linalg.eig ──
    w_np = np.sort_complex(np.linalg.eigvals(A))
    w_my = np.sort_complex(lambdas)
    err = np.max(np.abs(w_np - w_my))
    print(f"\nSai số giá trị riêng so với numpy.linalg.eig: {_sci(err)}"
          f"  →  {'ĐÚNG ✓' if err < 1e-6 else 'SAI ✗'}")
```

# Câu 45:
Viết thuật toán tìm giá trị riêng trội nhất của ma trận A và véc-tơ 
riêng tương ứng bằng phương pháp lũy thừa. Áp dụng cho một ma trận 
đối xứng cấp 8.

## Phương pháp lũy thừa (Power Method)

### Bài toán

Cho ma trận $A \in \text{mat}(n, \mathbb{R})$. Tìm **giá trị riêng trội nhất** $\lambda_1$ (giá trị riêng có trị tuyệt đối lớn nhất) và **vector riêng** tương ứng $v_1$.

**Giả thiết:** $A$ có $n$ giá trị riêng thoả

$$|\lambda_1| > |\lambda_2| \geq \cdots \geq |\lambda_n|$$

($\lambda_1$ tách hẳn — gọi là giá trị riêng **trội đơn**). Với $A$ **đối xứng thực**, mọi giá trị riêng đều thực và có cơ sở vector riêng trực chuẩn, nên phương pháp đặc biệt ổn định.

---

### Cơ sở lý thuyết

Khai triển vector khởi đầu $x^{(0)}$ theo cơ sở vector riêng $\{v_1, \ldots, v_n\}$:

$$x^{(0)} = c_1 v_1 + c_2 v_2 + \cdots + c_n v_n, \quad (c_1 \neq 0)$$

Nhân liên tiếp với $A$:

$$A^k x^{(0)} = c_1 \lambda_1^k v_1 + c_2 \lambda_2^k v_2 + \cdots = \lambda_1^k \left( c_1 v_1 + \sum_{i=2}^{n} c_i \left(\tfrac{\lambda_i}{\lambda_1}\right)^k v_i \right)$$

Vì $\left|\tfrac{\lambda_i}{\lambda_1}\right| < 1$ với $i \geq 2$, các số hạng sau **tắt dần**, nên:

$$\frac{A^k x^{(0)}}{\lambda_1^k} \;\xrightarrow{k \to \infty}\; c_1 v_1$$

Hướng của $A^k x^{(0)}$ hội tụ về vector riêng trội $v_1$. **Tốc độ hội tụ** tuyến tính với hệ số $\left|\tfrac{\lambda_2}{\lambda_1}\right|$ — tỉ số này càng nhỏ, hội tụ càng nhanh.

**Ước lượng giá trị riêng.** Sau khi có xấp xỉ vector riêng, dùng **thương Rayleigh** (tối ưu cho ma trận đối xứng):

$$\lambda_1 \approx \frac{x^T A x}{x^T x}$$

---

### Thuật toán

**Đầu vào:** Ma trận $A$, vector khởi đầu $x^{(0)} \neq 0$, sai số $\varepsilon$, số vòng lặp tối đa $N$

**Đầu ra:** Giá trị riêng trội $\lambda_1$ và vector riêng $v_1$ (chuẩn hoá)

---

**Bước 1 — Khởi tạo**

&emsp;Chuẩn hoá $x \leftarrow x^{(0)} / \|x^{(0)}\|$. Đặt $\lambda_{\text{cũ}} \leftarrow 0$, $k \leftarrow 0$.

**Bước 2 — Vòng lặp:** lặp khi $k < N$

&emsp;**2.1.** Nhân: $y \leftarrow A x$

&emsp;**2.2.** Ước lượng giá trị riêng (thương Rayleigh): $\lambda \leftarrow x^T y \;/\; x^T x$

&emsp;**2.3.** Chuẩn hoá lại: $x \leftarrow y / \|y\|$

&emsp;**2.4.** Nếu $|\lambda - \lambda_{\text{cũ}}| < \varepsilon$: **dừng**, trả về $(\lambda, x)$

&emsp;**2.5.** $\lambda_{\text{cũ}} \leftarrow \lambda$, $\;k \leftarrow k + 1$

**Bước 3 — Kết thúc**

&emsp;Nếu $k = N$ mà chưa hội tụ: thông báo *"Vượt quá số vòng lặp tối đa"*.

---

### Lưu ý

- **Chuẩn hoá mỗi bước là bắt buộc:** Nếu không, $\|A^k x^{(0)}\| \sim |\lambda_1|^k$ sẽ **tràn số** (khi $|\lambda_1|>1$) hoặc **triệt tiêu** (khi $|\lambda_1|<1$). Chuẩn hoá giữ vector trong tầm kiểm soát mà không đổi hướng.

- **Chọn $x^{(0)}$:** Cần $c_1 \neq 0$ (thành phần theo $v_1$ khác 0). Vector ngẫu nhiên hầu như luôn thoả; chỉ xui rủi khi $x^{(0)} \perp v_1$.

- **Giá trị riêng trội âm hoặc phức:** Nếu $\lambda_1 < 0$, vector đổi dấu mỗi bước ($x^{(k)}$ dao động) — thương Rayleigh vẫn hội tụ đúng dấu. Nếu $|\lambda_1| = |\lambda_2|$ (ví dụ cặp phức liên hợp trội), phương pháp lũy thừa **không hội tụ** — cần biến thể khác.

- **Vì sao chọn ma trận đối xứng:** $A = A^T$ ⟹ giá trị riêng thực, vector riêng trực giao, và thương Rayleigh có sai số bậc hai $O(\|x - v_1\|^2)$ — chính xác hơn hẳn cách ước lượng theo tỉ số thành phần.

- **Mở rộng:** Kết hợp **phép dịch** $A - \sigma I$ (shift) hoặc **lũy thừa nghịch đảo** $(A - \sigma I)^{-1}$ để tìm giá trị riêng **nhỏ nhất** hoặc **gần một giá trị $\sigma$ cho trước**.

```python
import numpy as np


# ── Định dạng số khoa học kiểu 1.23×10⁻⁵ ─────────────────────────
def _sci(x, sig=4):
    if abs(x) < 1e-14:
        return "0"
    s = f"{x:.{sig}e}"
    m, e = s.split("e")
    exp = int(e)
    sup = str(abs(exp)).translate(str.maketrans("0123456789", "⁰¹²³⁴⁵⁶⁷⁸⁹"))
    sign = "⁻" if exp < 0 else ""
    return f"{m}×10{sign}{sup}"


# ── Phương pháp lũy thừa ─────────────────────────────────────────
def power_method(A, x0=None, eps=1e-10, max_iter=1000, log=8):
    """
    Tìm giá trị riêng trội nhất và vector riêng bằng phương pháp lũy thừa.

    Tham số:
        A       : ma trận (n x n)
        x0      : vector khởi đầu (mặc định = [1,1,...,1])
        eps     : sai số cho phép của giá trị riêng
        max_iter: số vòng lặp tối đa
        log     : in ra log vài lần lặp đầu (0 = tắt)

    Trả về:
        lam  : giá trị riêng trội
        x    : vector riêng (chuẩn hoá)
        k    : số vòng lặp
        info : "ok" / thông báo
    """
    A = np.array(A, dtype=float)
    n = len(A)
    x = np.ones(n) if x0 is None else np.array(x0, dtype=float)
    x = x / np.linalg.norm(x)

    lam_old = 0.0
    if log:
        print(f"{'k':>3}  {'λ (Rayleigh)':>16}  {'|λ - λ_cũ|':>14}")

    for k in range(1, max_iter + 1):
        y = A @ x
        lam = x @ y / (x @ x)           # thương Rayleigh
        x = y / np.linalg.norm(y)

        delta = abs(lam - lam_old)
        if log and k <= log:
            print(f"{k:>3}  {lam:>16.10f}  {_sci(delta):>14}")

        if delta < eps:
            return lam, x, k, "ok"
        lam_old = lam

    return lam, x, max_iter, "Vượt quá số vòng lặp tối đa"


# ====== Áp dụng với ma trận đối xứng cấp 8 ======
if __name__ == "__main__":
    # Sinh ma trận đối xứng cấp 8 từ một ma trận cố định: A = (M + M^T)/2
    M = np.array([
        [12,  2,  1,  3,  1,  2,  1,  2],
        [ 2, 10,  2,  1,  2,  1,  3,  1],
        [ 1,  2, 14,  2,  1,  3,  1,  2],
        [ 3,  1,  2, 11,  2,  1,  2,  1],
        [ 1,  2,  1,  2, 13,  2,  1,  3],
        [ 2,  1,  3,  1,  2, 15,  2,  1],
        [ 1,  3,  1,  2,  1,  2, 10,  2],
        [ 2,  1,  2,  1,  3,  1,  2, 16],
    ], dtype=float)
    A = (M + M.T) / 2                    # đảm bảo đối xứng chính xác
    n = len(A)

    print("Ma trận A đối xứng?  ", np.allclose(A, A.T), "\n")

    lam, x, k, info = power_method(A, eps=1e-12)
    print(f"\nTrạng thái : {info}")
    print(f"Số vòng lặp: {k}")
    print(f"\nGiá trị riêng trội  λ₁ = {lam:.10f}")
    print("Vector riêng v₁:")
    print("  [" + "  ".join(f"{v:8.5f}" for v in x) + "]")

    # ── Kiểm tra ──
    res = np.linalg.norm(A @ x - lam * x)
    print(f"\n=== Kiểm tra ===")
    print(f"  ||A·v₁ - λ₁·v₁||          = {_sci(res)}")

    eigvals = np.linalg.eigvalsh(A)     # dùng eigvalsh cho ma trận đối xứng
    lam_true = eigvals[np.argmax(np.abs(eigvals))]
    print(f"  λ₁ (numpy.eigvalsh)       = {lam_true:.10f}")
    print(f"  Sai số so với numpy       = {_sci(abs(lam - lam_true))}"
          f"  →  {'ĐÚNG ✓' if abs(lam - lam_true) < 1e-8 else 'SAI ✗'}")

    ratio = abs(np.sort(np.abs(eigvals))[-2] / lam_true)
    print(f"\n  Tỉ số hội tụ |λ₂/λ₁|      = {ratio:.4f}  (càng nhỏ hội tụ càng nhanh)")
    print(f"  Phổ giá trị riêng (|·| giảm dần):")
    for ev in sorted(eigvals, key=lambda t: -abs(t)):
        print(f"    {ev:10.5f}")
```

# Câu 46:
Viết thuật toán cho phương pháp xuống thang tìm giá trị riêng trội thứ hai.  

## Phương pháp xuống thang (Deflation)

### Bài toán

Đã biết giá trị riêng trội nhất $\lambda_1$ và vector riêng $v_1$ (từ phương pháp lũy thừa — Câu 45). Tìm giá trị riêng **trội thứ hai** $\lambda_2$ và vector riêng tương ứng.

**Ý tưởng:** "Loại bỏ" thành phần ứng với $\lambda_1$ khỏi ma trận, tạo ma trận mới $A_2$ có cùng phổ nhưng $\lambda_1$ bị thay bằng $0$. Khi đó $\lambda_2$ trở thành giá trị riêng **trội nhất** của $A_2$, và ta lại dùng phương pháp lũy thừa.

---

### Cơ sở lý thuyết — Xuống thang Hotelling (ma trận đối xứng)

Với $A$ **đối xứng thực**, các vector riêng trực giao. Chọn $v_1$ đã **chuẩn hoá** ($\|v_1\| = 1$), lập:

$$A_2 = A - \lambda_1 \, v_1 v_1^T$$

**Kiểm chứng phổ:** với mọi vector riêng $v_i$ của $A$ (do trực giao $v_1^T v_i = \delta_{1i}$):

$$A_2 v_i = A v_i - \lambda_1 v_1 (v_1^T v_i) =
\begin{cases}
\lambda_1 v_1 - \lambda_1 v_1 = 0 & i = 1 \\
\lambda_i v_i - 0 = \lambda_i v_i & i \neq 1
\end{cases}$$

Vậy $A_2$ giữ nguyên mọi cặp $(\lambda_i, v_i)$ với $i \geq 2$, chỉ **triệt tiêu** $\lambda_1 \to 0$. Giờ giá trị riêng trội nhất của $A_2$ chính là $\lambda_2$.

> **Trường hợp không đối xứng:** vector riêng không trực giao, phải dùng xuống thang **Wielandt** (cần thêm vector riêng trái $w_1$): $A_2 = A - \lambda_1 \dfrac{v_1 w_1^T}{w_1^T v_1}$. Phần này tập trung vào ma trận đối xứng nên dùng Hotelling.

---

### Thuật toán

**Đầu vào:** Ma trận đối xứng $A$, cặp trội $(\lambda_1, v_1)$ đã chuẩn hoá

**Đầu ra:** Giá trị riêng trội thứ hai $\lambda_2$ và vector riêng $v_2$

---

**Bước 1 — Xuống thang**

&emsp;Lập ma trận đã khử trội: $\;A_2 \leftarrow A - \lambda_1 \, v_1 v_1^T$

**Bước 2 — Lũy thừa trên $A_2$**

&emsp;Áp dụng phương pháp lũy thừa (Câu 45) cho $A_2$ để tìm $(\lambda_2, v_2)$ — chính là giá trị riêng trội nhất của $A_2$.

**Bước 3 — Kiểm tra**

&emsp;Xác nhận $\|A v_2 - \lambda_2 v_2\| \approx 0$ **trên ma trận gốc $A$** (không phải $A_2$), vì $v_2$ cũng là vector riêng của $A$.

---

### Lưu ý

- **Vì sao kiểm tra trên $A$ chứ không $A_2$:** $v_2$ là vector riêng chung của cả $A$ và $A_2$ (chỉ khác giá trị riêng ở hướng $v_1$). Kiểm tra $A v_2 = \lambda_2 v_2$ xác nhận kết quả đúng với bài toán gốc.

- **Tích luỹ sai số:** Mỗi lần xuống thang, sai số của $(\lambda_1, v_1)$ "rò rỉ" vào $A_2$. Nếu lặp xuống thang nhiều lần (Câu 47), sai số **tích luỹ dần** — nên tính $\lambda_1, v_1$ càng chính xác càng tốt trước khi xuống thang.

- **Yêu cầu trực giao:** Xuống thang Hotelling **chỉ đúng** khi $v_1$ thực sự trực giao với các vector riêng còn lại — điều được bảo đảm bởi tính đối xứng của $A$. Với ma trận không đối xứng, công thức này cho kết quả sai.

- **Tách trội thứ hai:** Để lũy thừa trên $A_2$ hội tụ, cần $|\lambda_2| > |\lambda_3|$. Nếu $\lambda_2, \lambda_3$ gần nhau, hội tụ chậm.

```python
import numpy as np


# ── Định dạng số khoa học kiểu 1.23×10⁻⁵ ─────────────────────────
def _sci(x, sig=4):
    if abs(x) < 1e-14:
        return "0"
    s = f"{x:.{sig}e}"
    m, e = s.split("e")
    exp = int(e)
    sup = str(abs(exp)).translate(str.maketrans("0123456789", "⁰¹²³⁴⁵⁶⁷⁸⁹"))
    sign = "⁻" if exp < 0 else ""
    return f"{m}×10{sign}{sup}"


# ── Phương pháp lũy thừa (như Câu 45) ────────────────────────────
def power_method(A, x0=None, eps=1e-12, max_iter=2000):
    """Trả về (λ trội, vector riêng chuẩn hoá, số vòng lặp)."""
    A = np.array(A, dtype=float)
    n = len(A)
    x = np.ones(n) if x0 is None else np.array(x0, dtype=float)
    x = x / np.linalg.norm(x)
    lam_old = 0.0
    for k in range(1, max_iter + 1):
        y = A @ x
        lam = x @ y / (x @ x)               # thương Rayleigh
        x = y / np.linalg.norm(y)
        if abs(lam - lam_old) < eps:
            return lam, x, k
        lam_old = lam
    return lam, x, max_iter


# ── Xuống thang Hotelling tìm giá trị riêng trội thứ hai ──────────
def second_eigen(A, eps=1e-12):
    """
    Câu 46: tìm (λ₂, v₂) — giá trị riêng trội thứ hai.

    Bước 1: lũy thừa trên A       → (λ₁, v₁)
    Bước 2: A₂ = A - λ₁ v₁ v₁ᵀ    (xuống thang)
    Bước 3: lũy thừa trên A₂      → (λ₂, v₂)
    """
    A = np.array(A, dtype=float)
    lam1, v1, k1 = power_method(A, eps=eps)
    A2 = A - lam1 * np.outer(v1, v1)          # ma trận đã khử trội λ₁
    lam2, v2, k2 = power_method(A2, eps=eps)
    return lam1, v1, lam2, v2, k1, k2


# ====== Áp dụng với ma trận đối xứng cấp 8 ======
if __name__ == "__main__":
    M = np.array([
        [12,  2,  1,  3,  1,  2,  1,  2],
        [ 2, 10,  2,  1,  2,  1,  3,  1],
        [ 1,  2, 14,  2,  1,  3,  1,  2],
        [ 3,  1,  2, 11,  2,  1,  2,  1],
        [ 1,  2,  1,  2, 13,  2,  1,  3],
        [ 2,  1,  3,  1,  2, 15,  2,  1],
        [ 1,  3,  1,  2,  1,  2, 10,  2],
        [ 2,  1,  2,  1,  3,  1,  2, 16],
    ], dtype=float)
    A = (M + M.T) / 2

    lam1, v1, lam2, v2, k1, k2 = second_eigen(A)

    print(f"λ₁ (trội nhất)     = {lam1:.10f}   ({k1} vòng lũy thừa trên A)")
    print(f"λ₂ (trội thứ hai)  = {lam2:.10f}   ({k2} vòng lũy thừa trên A₂)")
    print("\nVector riêng v₂:")
    print("  [" + "  ".join(f"{v:8.5f}" for v in v2) + "]")

    # ── Kiểm tra ──
    res = np.linalg.norm(A @ v2 - lam2 * v2)         # kiểm tra trên A GỐC
    print("\n=== Kiểm tra ===")
    print(f"  ||A·v₂ - λ₂·v₂||       = {_sci(res)}   (trên ma trận gốc A)")
    print(f"  v₁ ⊥ v₂ ?  |v₁·v₂|     = {_sci(abs(v1 @ v2))}   (≈0 ⇒ trực giao)")

    eigvals = np.linalg.eigvalsh(A)
    lam2_true = sorted(eigvals, key=lambda t: -abs(t))[1]
    print(f"  λ₂ (numpy.eigvalsh)   = {lam2_true:.10f}")
    print(f"  Sai số so với numpy   = {_sci(abs(lam2 - lam2_true))}"
          f"  →  {'ĐÚNG ✓' if abs(lam2 - lam2_true) < 1e-8 else 'SAI ✗'}")
```

# Câu 47:
Viết thuật toán tìm tất cả các giá trị riêng và véc-tơ riêng tương 
ứng của ma trận A bằng phương pháp lũy thừa và phương pháp xuống 
thang.

## Tất cả giá trị riêng bằng Lũy thừa + Xuống thang

### Bài toán

Cho ma trận **đối xứng** $A$. Tìm **toàn bộ** $n$ cặp giá trị riêng — vector riêng $(\lambda_i, v_i)$, bằng cách kết hợp hai phương pháp đã có:

- **Lũy thừa** (Câu 45) — tìm cặp trội nhất của ma trận hiện tại.
- **Xuống thang Hotelling** (Câu 46) — khử cặp vừa tìm để "lộ" cặp trội kế tiếp.

**Ý tưởng:** Lặp lại xen kẽ hai bước trên $n$ lần, mỗi lần bóc ra một cặp riêng theo thứ tự $|\lambda_1| \geq |\lambda_2| \geq \cdots \geq |\lambda_n|$.

---

### Cơ sở lý thuyết

Đặt $A^{(1)} = A$. Ở vòng thứ $i$:

$$\big(\lambda_i, v_i\big) = \text{PowerMethod}\big(A^{(i)}\big), \qquad A^{(i+1)} = A^{(i)} - \lambda_i \, v_i v_i^T$$

Sau $i$ lần khử, ma trận $A^{(i+1)}$ có phổ $\{0, \ldots, 0, \lambda_{i+1}, \ldots, \lambda_n\}$ (các giá trị riêng đã bóc bị đưa về $0$). Vì $A$ đối xứng nên mọi $v_i$ trực giao, phép khử Hotelling **chính xác** ở mỗi bước, và cuối cùng thu được **phân tích phổ** đầy đủ:

$$A = \sum_{i=1}^{n} \lambda_i \, v_i v_i^T = Q \Lambda Q^T$$

với $Q = [v_1 \, v_2 \, \cdots \, v_n]$ trực giao và $\Lambda = \text{diag}(\lambda_1, \ldots, \lambda_n)$.

---

### Thuật toán

**Đầu vào:** Ma trận đối xứng $A$ cấp $n$, sai số $\varepsilon$

**Đầu ra:** Toàn bộ $(\lambda_i, v_i)$, $i = 1, \ldots, n$

---

**Bước 1 — Khởi tạo**

&emsp;$A^{(1)} \leftarrow A$; danh sách kết quả rỗng.

**Bước 2 — Vòng lặp bóc phổ:** với $i = 1, 2, \ldots, n$

&emsp;**2.1.** $(\lambda_i, v_i) \leftarrow$ **PowerMethod**$(A^{(i)})$ — cặp trội nhất hiện tại.

&emsp;**2.2.** Lưu $(\lambda_i, v_i)$.

&emsp;**2.3.** Xuống thang: $A^{(i+1)} \leftarrow A^{(i)} - \lambda_i \, v_i v_i^T$

**Bước 3 — Kiểm tra**

&emsp;Với mỗi cặp: $\|A v_i - \lambda_i v_i\| \approx 0$.

&emsp;Kiểm tra tổng thể: $Q$ trực chuẩn ($Q^T Q \approx I$) và tái tạo $\|Q \Lambda Q^T - A\| \approx 0$.

---

### Lưu ý

- **Sai số tích luỹ:** Đây là nhược điểm chính. Cặp $(\lambda_1, v_1)$ có sai số nhỏ, nhưng sai số đó truyền vào $A^{(2)}$, rồi $A^{(3)}$… nên **các giá trị riêng cuối kém chính xác hơn** các giá trị riêng đầu. Với $n$ lớn, nên dùng thuật toán chuyên dụng (QR) thay vì xuống thang lặp.

- **Chỉ hợp với ma trận đối xứng:** Toàn bộ lập luận dựa vào tính trực giao của vector riêng. Ma trận không đối xứng cần xuống thang Wielandt và xử lý giá trị riêng phức — phức tạp hơn nhiều.

- **Giá trị riêng bội / gần nhau:** Nếu $|\lambda_i| \approx |\lambda_{i+1}|$, phương pháp lũy thừa ở bước đó hội tụ **rất chậm** (tỉ số $|\lambda_{i+1}/\lambda_i| \to 1$). Đây là lý do một vài giá trị riêng ở giữa cần nhiều vòng lặp.

- **So sánh với Danilevski (Câu 44):** Danilevski cho *toàn bộ* phổ qua đa thức đặc trưng trong $O(n^3)$ một lần, kể cả giá trị riêng phức; còn lũy thừa + xuống thang chỉ hợp ma trận đối xứng nhưng **không cần** giải đa thức bậc cao (vốn nhạy sai số khi $n$ lớn).

```python
import numpy as np


# ── Định dạng số khoa học kiểu 1.23×10⁻⁵ ─────────────────────────
def _sci(x, sig=4):
    if abs(x) < 1e-14:
        return "0"
    s = f"{x:.{sig}e}"
    m, e = s.split("e")
    exp = int(e)
    sup = str(abs(exp)).translate(str.maketrans("0123456789", "⁰¹²³⁴⁵⁶⁷⁸⁹"))
    sign = "⁻" if exp < 0 else ""
    return f"{m}×10{sign}{sup}"


# ── Phương pháp lũy thừa (như Câu 45) ────────────────────────────
def power_method(A, x0=None, eps=1e-12, max_iter=5000):
    """Trả về (λ trội, vector riêng chuẩn hoá, số vòng lặp)."""
    A = np.array(A, dtype=float)
    n = len(A)
    x = np.ones(n) if x0 is None else np.array(x0, dtype=float)
    x = x / np.linalg.norm(x)
    lam_old = 0.0
    for k in range(1, max_iter + 1):
        y = A @ x
        lam = x @ y / (x @ x)               # thương Rayleigh
        x = y / np.linalg.norm(y)
        if abs(lam - lam_old) < eps:
            return lam, x, k
        lam_old = lam
    return lam, x, max_iter


# ── Tất cả (λ, v) bằng lũy thừa + xuống thang lặp ────────────────
def all_eigen_deflation(A, eps=1e-12):
    """
    Câu 47: bóc lần lượt từng cặp riêng trội nhất rồi xuống thang.

    Trả về:
        lambdas : mảng n giá trị riêng (theo thứ tự |λ| giảm dần)
        V       : ma trận n×n, cột i là vector riêng của lambdas[i]
        iters   : số vòng lũy thừa mỗi lần
    """
    A = np.array(A, dtype=float)
    n = len(A)
    Ak = A.copy()
    lambdas, vectors, iters = [], [], []
    for _ in range(n):
        lam, v, k = power_method(Ak, eps=eps)
        lambdas.append(lam)
        vectors.append(v)
        iters.append(k)
        Ak = Ak - lam * np.outer(v, v)        # xuống thang Hotelling
    return np.array(lambdas), np.array(vectors).T, iters


# ====== Áp dụng với ma trận đối xứng cấp 8 ======
if __name__ == "__main__":
    M = np.array([
        [12,  2,  1,  3,  1,  2,  1,  2],
        [ 2, 10,  2,  1,  2,  1,  3,  1],
        [ 1,  2, 14,  2,  1,  3,  1,  2],
        [ 3,  1,  2, 11,  2,  1,  2,  1],
        [ 1,  2,  1,  2, 13,  2,  1,  3],
        [ 2,  1,  3,  1,  2, 15,  2,  1],
        [ 1,  3,  1,  2,  1,  2, 10,  2],
        [ 2,  1,  2,  1,  3,  1,  2, 16],
    ], dtype=float)
    A = (M + M.T) / 2
    n = len(A)

    lambdas, V, iters = all_eigen_deflation(A)

    print("Tất cả giá trị riêng & vector riêng (|λ| giảm dần):")
    print("─" * 78)
    print(f"{'i':>2}  {'λ_i':>14}  {'#vòng':>6}  {'||A·v-λv||':>13}  {'sai số vs numpy':>16}")
    print("─" * 78)
    ev_true = np.sort(np.linalg.eigvalsh(A))[::-1]      # thực, giảm dần
    max_res = 0.0
    for i in range(n):
        lam, v, k = lambdas[i], V[:, i], iters[i]
        res = np.linalg.norm(A @ v - lam * v)
        max_res = max(max_res, res)
        err = abs(lam - ev_true[i])
        print(f"{i+1:>2}  {lam:>14.8f}  {k:>6}  {_sci(res):>13}  {_sci(err):>16}")
    print("─" * 78)

    print("\nVector riêng (mỗi cột 1 vector, chuẩn hoá):")
    for r in range(n):
        print("  " + "  ".join(f"{V[r, c]:8.4f}" for c in range(n)))

    # ── Kiểm tra tổng thể ──
    Q = V
    orth  = np.max(np.abs(Q.T @ Q - np.eye(n)))            # trực chuẩn?
    recon = np.max(np.abs(Q @ np.diag(lambdas) @ Q.T - A)) # tái tạo A?
    print("\n=== Kiểm tra tổng thể ===")
    print(f"  Residual lớn nhất  max‖A·v-λv‖ = {_sci(max_res)}")
    print(f"  Độ lệch trực chuẩn   |QᵀQ - I| = {_sci(orth)}")
    print(f"  Tái tạo phổ        |QΛQᵀ - A|  = {_sci(recon)}"
          f"  →  {'ĐÚNG ✓' if recon < 1e-6 else 'SAI ✗'}")
```
