# Sai số và số gần đúng

## 1. Số gần đúng và sai số

Khi $a$ là số gần đúng của giá trị đúng $a^*$:

**Sai số tuyệt đối:**

$$\Delta a = |a^* - a|$$

Trong thực tế không biết $a^*$, nên dùng **cận sai số tuyệt đối** $\Delta$ sao cho $|a^* - a| \leq \Delta$. Khi đó viết:

$$a^* = a \pm \Delta$$

**Sai số tương đối:**

$$\delta a = \frac{\Delta a}{|a|}$$

Thường tính bằng phần trăm: $\delta a \times 100\%$.

---

## 2. Chữ số đáng tin

Một chữ số của số gần đúng $a$ là **đáng tin** nếu sai số tuyệt đối không vượt quá nửa đơn vị ở hàng chứa chữ số đó.

**Công thức:** Nếu chữ số đáng tin đầu tiên nằm ở hàng $10^k$ và $a$ có $m$ chữ số đáng tin, thì:

$$\Delta a \leq 0.5 \times 10^{k - m + 1}$$

**Quy tắc ngược (biết $\Delta a$, tìm $m$):** Tìm hàng $10^n$ nhỏ nhất sao cho $\Delta a \leq 0.5 \times 10^n$, rồi đếm từ chữ số đầu tiên đến hàng $10^n$.

**Ví dụ:** $a = 3.1416$, $\Delta a = 3 \times 10^{-5}$.

- $3 \times 10^{-5} \leq 0.5 \times 10^{-4}$ → hàng cuối đáng tin là $10^{-4}$
- Chữ số đầu tiên ở hàng $10^0$
- Số chữ số đáng tin: từ hàng $10^0$ đến $10^{-4}$ → **5 chữ số** (3, 1, 4, 1, 6)

---

## 3. Quy tắc làm tròn

Khi làm tròn $a$ đến $n$ chữ số đáng tin, sai số do làm tròn:

$$\Delta_{\text{tròn}} \leq 0.5 \times 10^{k}$$

với $k$ là hàng của chữ số cuối giữ lại.

**Lưu ý:** Không giữ lại chữ số không đáng tin — nó tạo cảm giác chính xác giả.

---

## 4. Sai số của phép tính số học

| Phép tính | Sai số tuyệt đối | Sai số tương đối |
|---|---|---|
| $u = a \pm b$ | $\Delta u = \Delta a + \Delta b$ | — (dùng tuyệt đối) |
| $u = a \cdot b$ | — (dùng tương đối) | $\delta u = \delta a + \delta b$ |
| $u = a / b$ | — (dùng tương đối) | $\delta u = \delta a + \delta b$ |
| $u = a^n$ | — | $\delta u = n \cdot \delta a$ |

**Quy tắc nhớ:**

- Cộng / trừ → sai số **tuyệt đối** cộng lại
- Nhân / chia / lũy thừa → sai số **tương đối** cộng lại

## 5. Nguyên lý lan truyền sai số

### Phát biểu

Cho hàm $u = f(x_1, x_2, \ldots, x_n)$ với các đối số $x_i$ có sai số tuyệt đối $\Delta x_i$. Sai số tuyệt đối của $u$ được ước lượng bởi:

$$\Delta u \approx \sum_{i=1}^{n} \left|\frac{\partial f}{\partial x_i}\right| \cdot \Delta x_i$$

### Ý nghĩa từng thành phần

Mỗi số hạng $\left|\frac{\partial f}{\partial x_i}\right| \cdot \Delta x_i$ là **phần đóng góp sai số** của biến $x_i$ vào kết quả:

- $\left|\frac{\partial f}{\partial x_i}\right|$ — **hệ số nhạy**: đo mức độ $u$ thay đổi khi $x_i$ thay đổi
- $\Delta x_i$ — sai số của bản thân biến $x_i$
- Biến có đạo hàm riêng lớn → "nhạy" hơn → sai số nhỏ ở biến đó cũng gây ảnh hưởng lớn đến kết quả

### Ví dụ: Thể tích hình cầu $V = \frac{\pi d^3}{6}$

Có hai biến: $\pi$ và $d$. Tính đạo hàm riêng:

$$\frac{\partial V}{\partial d} = \frac{\pi d^2}{2}, \quad \frac{\partial V}{\partial \pi} = \frac{d^3}{6}$$

Áp dụng công thức lan truyền:

$$\Delta V = \left|\frac{\partial V}{\partial d}\right| \cdot \Delta d + \left|\frac{\partial V}{\partial \pi}\right| \cdot \Delta\pi = \frac{\pi d^2}{2} \cdot \Delta d + \frac{d^3}{6} \cdot \Delta\pi$$

Thay $d \approx 1.64$, $\pi = 3.1415926$, $\Delta\pi = 0.5 \times 10^{-7}$:

- Phần do $d$: $\frac{3.1416 \times 1.64^2}{2} \cdot \Delta d = 4.226 \cdot \Delta d$
- Phần do $\pi$: $\frac{1.64^3}{6} \times 0.5 \times 10^{-7} \approx 3.68 \times 10^{-8}$ (rất nhỏ)

→ Biến $d$ nhạy hơn nhiều so với $\pi$, nên sai số của $d$ chi phối sai số của $V$.

## 6. Nguyên lý ảnh hưởng đều của sai số

### Bài toán

Biết sai số cho phép $\Delta u$ của kết quả. Cần xác định sai số $\Delta x_i$ cho **từng biến đầu vào** sao cho hợp lý.

### Phát biểu

Phân bổ sai số sao cho **mỗi biến đóng góp sai số ngang nhau** vào kết quả:

$$\left|\frac{\partial f}{\partial x_1}\right| \Delta x_1 = \left|\frac{\partial f}{\partial x_2}\right| \Delta x_2 = \ldots = \left|\frac{\partial f}{\partial x_n}\right| \Delta x_n$$

Vì tổng các phần bằng $\Delta u$ và có $n$ phần bằng nhau, mỗi phần bằng $\frac{\Delta u}{n}$. Từ đó:

$$\boxed{\Delta x_i = \frac{\Delta u}{n \cdot \left|\frac{\partial f}{\partial x_i}\right|}}$$

### Tại sao phải chia đều?

Nếu phân bổ **không đều**, ví dụ biến $x_1$ chiếm 90% sai số, biến $x_2$ chỉ 10%, thì:

- Ta đang "lãng phí" độ chính xác ở $x_2$ (tính chính xác hơn mức cần thiết → tốn công)
- Trong khi $x_1$ quá thô → chi phối gần như toàn bộ sai số

Chia đều giúp **không biến nào phải chính xác quá mức cần thiết, cũng không biến nào quá thô** so với các biến còn lại.

### Ví dụ: Tiếp tục với $V = \frac{\pi d^3}{6}$

Yêu cầu $V$ có 6 chữ số đáng tin, $V \approx 2.3$ → $\Delta V \leq 0.5 \times 10^{-5} = 5 \times 10^{-6}$.

Có $n = 2$ biến ($\pi$ và $d$), mỗi biến đóng góp $\frac{\Delta V}{2} = 2.5 \times 10^{-6}$.

**Sai số cần thiết cho $d$:**

$$\frac{\pi d^2}{2} \cdot \Delta d = 2.5 \times 10^{-6} \implies \Delta d = \frac{2.5 \times 10^{-6}}{\frac{3.1416 \times 1.64^2}{2}} = \frac{2.5 \times 10^{-6}}{4.226} \approx 5.9 \times 10^{-7}$$

→ Đây là $\varepsilon$ cho phương pháp lặp tìm $d$.

**Sai số cần thiết cho $\pi$:**

$$\frac{d^3}{6} \cdot \Delta\pi = 2.5 \times 10^{-6} \implies \Delta\pi = \frac{2.5 \times 10^{-6}}{\frac{1.64^3}{6}} = \frac{2.5 \times 10^{-6}}{0.735} \approx 3.4 \times 10^{-6}$$

→ Cần $\pi$ chính xác đến $\Delta\pi \leq 3.4 \times 10^{-6}$, tức khoảng 6 chữ số sau dấu phẩy.

### Kiểm chứng

$$\Delta V = 4.226 \times 5.9 \times 10^{-7} + 0.735 \times 3.4 \times 10^{-6} \approx 2.5 \times 10^{-6} + 2.5 \times 10^{-6} = 5 \times 10^{-6} \leq \Delta V_{\text{yêu cầu}} \checkmark$$

Hai phần đóng góp bằng nhau → nguyên lý ảnh hưởng đều được thoả mãn.
